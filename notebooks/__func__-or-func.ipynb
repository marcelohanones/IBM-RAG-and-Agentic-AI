# .func vs __func__

## Step 1 — check what kind of object you're holding.



> se for `<class 'functools.partial'>` →  Use  `.func` <br> <br> <br> 
> se for `<class 'method'>` ou `<class 'builtin_function_or_method'> (bound method, incl. classmethods)` → use  `__func__`

```python
import inspect
import typing

from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts.few_shot_with_templates import FewShotPromptWithTemplates
from langchain_text_splitters import RecursiveCharacterTextSplitter
from pydantic import BaseModel
```

```python
_ = JsonOutputParser()
target = _.get_format_instructions   # note: parser., not JsonOutputParser.

(
    print(target), 
    type(target)
    )
```

```python
(
hasattr(target, 'func') ,   # True only for functools.partial ,
hasattr(target, '__func__')   # True only for bound methods)
)
```



|  Callable | Use  | Gets you  |
| --- | --- | --- |
| functools.partial(f, ...)	 | p.func | the original wrapped function f |
| Class.classmethod / instance.method | m.\_\_func\_\_ | the plain underlying function, unbound |
